Getting started with Langchain and OpenAI

We will learn how to :

-> Setup langchain,langsmith,langserve

-> use basic components of langchain : prompt templates, models, and o/p parser

-> build a simple application with langchain

-> trace your application with langsmith

-> serve your application with langserve


```python
import os
from dotenv import load_dotenv
load_dotenv()
# os.environ["HUGGING_FACE_API_KEY"]=os.getenv("HUGGING_FACE_API_KEY")
os.environ["LANGCHAIN_API_KEY"] = os.getenv("LANGCHAIN_API_KEY")
os.environ["LANGCHAIN_TRACING_V2"] = "true"
os.environ["LANGCHAIN_PROJECT"]=os.getenv("LANGCHAIN_PROJECT")
```

```python
from langchain_community.chat_models import ChatOllama


llm = ChatOllama(
    model="gemma3:1b",   # or "llama3", "mistral", etc (must be pulled in Ollama)
    temperature=0.2,
)


```

```python
# 1️⃣ What is ChatOllama?

# ChatOllama is a LangChain class.

# It’s just a Python wrapper that lets LangChain talk to the Ollama server running on your machine.

# You don’t “download” ChatOllama as a model — you use it to call a model that Ollama is hosting.
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
prompt = ChatPromptTemplate.from_messages(
    [
        ("system", "You are an expert AI Engineer. Provide me answers based on the questions"),
        ("user", "{input}")
    ]
)
```

```python
output_parser = StrOutputParser()
```

```python
chain = prompt | llm | output_parser
```

```python
response = chain.invoke({"input": "What is generative AI?"})
print(response)
```
