```python
# This application will translate text from english to other language.
# This would be built using lcel i.e Langchain expression language which is used to chain components together.
```

```python
import os
from dotenv import load_dotenv
load_dotenv()

groq_api_key = os.getenv("GROQ_API_KEY")
```

```python
from langchain_groq import ChatGroq
llm = ChatGroq(model="groq/compound",groq_api_key=groq_api_key)
# chatgroq is a wrapper around groq which is used to interact with groq
llm
```

```python
from langchain_core.messages import HumanMessage,SystemMessage
# whenever we provide any msg to our llm model we should specify that which
#  msg is provided by the human and which msg is an instruction to the llm model(system msg)
messages = [SystemMessage(content="Translate the following from english to french"),
            HumanMessage(content="Hello how are you?")]
result=llm.invoke(messages)
```

```python
result
```

```python
from langchain_core.output_parsers import StrOutputParser
# it is used to parse the output of the llm model , basically to display only the text part of the output
parser = StrOutputParser()
parsed_result = parser.invoke(result)
```

```python
parsed_result
```

```python
# using lcel we can chain the components together in a single expression as below
chain = llm|parser
chain.invoke(messages)
```

```python
# prompt templates
from langchain_core.prompts import ChatPromptTemplate
generic_template = "Translate the following into {language}:"
prompt = ChatPromptTemplate.from_messages([
    ("system",generic_template),("user","{text}")

])
```

```python
result=prompt.invoke({"language":"french","text":"Good morning, how are you?"})
```

```python
result.to_messages()
```

```python
chain = prompt|llm|parser
```

```python
chain.invoke({"language":"spanish","text":"What is your name?"})
```
