```python
# Every LLM model has its own limitation of context size, in order to take care of that we need to chunk the data accordingly.
```

```python
from langchain_community.document_loaders import PyPDFLoader
loader = PyPDFLoader('../RIA_Resume.pdf')
docs = loader.load()
docs
```

Text Splitting from Documents

```python
# Recursively splitting text by characters
from langchain_text_splitters import RecursiveCharacterTextSplitter
text_splitter = RecursiveCharacterTextSplitter(chunk_size=500,chunk_overlap=50)
# this means each chunk will be of 500 characters with an overlap of 50 characters between chunks
# overlap of 50 characters means that the last 50 characters of one chunk will be repeated at the beginning of the next chunk
# this is useful to maintain context between chunks
final_documents = text_splitter.split_documents(docs)
# we used split_documents as we are passing document object
# if we were passing plain text we would have used create_document method
```

```python
print(final_documents[0])
print(final_documents[1])
```
